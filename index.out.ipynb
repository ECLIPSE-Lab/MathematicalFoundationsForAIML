## 3 Week 1 Summary: Learning vs Data Analysis; Models, Loss Functions

- **Models:** Simplified representations for prediction/explanation (White, Grey, Black-Box).
- **Learning Types:** Supervised (Regression/Classification), Unsupervised, Reinforcement.
- **Loss Functions:** ML minimizes a loss function (e.g., MSE) to match predictions to targets.
- **Complexity:** Occam’s Razor prefers simpler models to avoid overfitting; regularization helps control it.

## 4 Week 2 Summary: Linear Algebra Refresher; Covariance, PCA/SVD

- **Eigendecomposition:** Finding eigenvectors (variance directions) and eigenvalues (variance amount) of a covariance matrix.
- **PCA:** Linear dimensionality reduction maximizing variance and compressing data.
- **Scree Plots:** Diagnostics to find intrinsic dimensionality (“elbow”).
- **Engineering Uses:** Denoising, anomaly detection, and compression (limited to linear manifolds).

## 5 Week 3 Summary: Regression as Loss Minimization

- **Supervised Framework:** Minimizing a cost function (MSE) to find optimal parameters.
- **Optimization:** Analytical (Ordinary Least Squares) vs. Iterative (Gradient Descent).
- **Basis Functions:** Expanding linear models to fit non-linear data using transformations (polynomials, splines).
- **Runge’s Phenomenon:** Overfitting risk with high-order global polynomials.

## 6 Week 4 Summary: Neural Networks Early: Neuron & Activations

- **Artificial Neuron:** Weighted sum plus a non-linear activation function.
- **Activations:** Non-linearity (ReLU, Sigmoid, Tanh) is essential for complex problem-solving.
- **Universal Approximation:** A single hidden layer can approximate any continuous function.
- **MLP Architecture:** Feed-forward network with input, hidden, and output layers.

## 7 Week 5 Summary: Backpropagation & Gradient Flow

- **Backpropagation:** Two-pass chain rule application to efficiently compute gradients.
- **Vanishing/Exploding Gradients:** Deep networks struggle with signal decay or instability.
- **ReLU:** Improves gradient flow for deep architectures by providing a constant positive gradient.
- **Jacobian Matrix:** Links differentiable components for system-wide optimization.

## 8 Week 6 Summary: Loss Landscapes & Optimization Behavior

- **Loss Landscape:** High-dimensional topography determining optimization success.
- **Curvature:** Hessian matrix dictates steepness; ill-conditioned landscapes slow descent.
- **Saddle Points:** Common traps in high dimensions that hinder optimizers.
- **Advanced Optimizers:** Momentum and adaptive learning rates (ADAM) navigate complex landscapes robustly.

## 9 Week 7 Summary: Generalization, Bias-Variance, Regularization

- **Generalization:** Ensuring accurate performance on unseen test data.
- **Bias-Variance Tradeoff:** Balancing underfitting (high bias) and overfitting (high variance).
- **Regularization:** L2 (Ridge), L1 (Lasso), and Dropout penalize complexity to improve robustness.
- **Validation:** Cross-validation is essential for unbiased hyperparameter tuning.

## 10 Week 8 Summary: Probabilistic View of Learning; Noise

- **Taxonomy of Uncertainty:** Aleatory (inherent noise) vs. Epistemic (lack of data).
- **Inference Views:** Frequentist (Maximum Likelihood) vs. Bayesian (Posterior updates).
- **Robustness:** Student’s t-distribution mitigates outlier impacts better than Gaussian.
- **Advanced UQ:** Stochastic enrichment and Mixture-Density Networks capture uncertainty.

## 11 Week 9 Summary: Representation Learning; Features vs. Learned Representations

- **Manifold Hypothesis:** High-dimensional data lies on a lower-dimensional non-linear manifold.
- **Autoencoders (AE):** Unsupervised neural networks compressing data into a latent bottleneck.
- **AE Architectures:** Encoders compress, decoders reconstruct; CNNs handle spatial data.
- **Applications:** Data compression, feature extraction, and anomaly detection.

## 12 Week 10 Summary: Latent Spaces; Embeddings

- **Latent Spaces:** Low-dimensional vectors where distance reflects semantic/physical similarity.
- **t-SNE & UMAP:** Non-linear visualization tools to map high-dimensional clusters to 2D/3D.
- **Crowding Problem:** Addressed by heavy-tailed distributions in lower dimensions.
- **Interpretability:** Mapping latent neurons to physical parameters builds trust.

## 13 Week 11 Summary: Unsupervised Learning Revisited (Objectives)

- **Unsupervised Paradigm:** Uncovering regimes and clusters in unlabeled process data.
- **K-Means & K-Medoids:** Hard clustering minimizing distance to centroids/prototypes.
- **Gaussian Mixture Models (GMM):** Probabilistic clustering with soft assignments (responsibilities).
- **EM Algorithm:** Iterative E-step and M-step to find maximum likelihood in latent variable models.

## 14 Week 12 Summary: Uncertainty in Predictions

- **Uncertainty Quantification (UQ):** Crucial for safe out-of-distribution engineering decisions.
- **Bayesian Predictive Distribution:** Integrating over parameter posteriors for robust variance estimates.
- **Evidence Framework:** Optimizing hyperparameters via Marginal Likelihood maximization.
- **Physics Constraints:** Embedding ODEs/PDEs reduces epistemic uncertainty in data-poor regions.

## 15 Week 13 Summary: Physics-informed & Constrained Learning

- **PINNs:** Integrating physical laws directly into the loss function to reduce data needs.
- **Data Enrichment:** Applying known mathematical transformations (FFT, derivatives).
- **Neural Integrators:** Using NNs with automatic differentiation as flexible differential equation solvers.
- **Scientific Trust:** Physics constraints act as powerful regularizers promoting Occam’s Razor.

## 16 Week 14 Summary: Explainability, Limits, Scientific Trust

- **Explainable AI (XAI):** Transitioning from black-box to transparent, justifiable decision-making.
- **Interpretation Tools:** Sensitivity analysis and deductive reasoning using ontologies.
- **Causality:** Moving from late detection to early prediction via causal process chains.
- **Limits:** Addressing data bias, extrapolation risks, and AI hallucinations while keeping experts in the loop.
